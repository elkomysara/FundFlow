````python
import requests
import json

BASE_URL = "http://localhost:8000/api/v1"

print("🧪 Testing ImaraFund Intelligent Matching System\n")

# Get first company
companies = requests.get(f"{BASE_URL}/companies?limit=1").json()
company = companies[0]

print(f"✅ Testing with: {company['company_name']}")
print(f"   Sector: {company['sector']}")
print(f"   Location: {company['nationality']}")
print(f"   Funding Need: ${company['funding_need_usd']:,}")

# Run matching algorithm
result = requests.post(f"{BASE_URL}/match/{company['id']}?top_n=3").json()

print(f"\n🎯 Found {result['total_matches_found']} matches above threshold\n")

# Display results
for i, match in enumerate(result['matches'], 1):
    print(f"Match #{i}: {match['program_name']}")
    print(f"Score: {match['match_score']}/100")
    
    breakdown = match['score_breakdown']
    print(f"Breakdown: Geography({breakdown['geographic']}/40) + "
          f"Sector({breakdown['sector']}/30) + "
          f"Funding({breakdown['amount_fit']}/20) + "
          f"Stage({breakdown['stage']}/10)")
    print(f"Website: {match['website']}\n")

# Show AI recommendation
if result.get('ai_recommendation'):
    print("🤖 AI Recommendation:")
    print(result['ai_recommendation'])
else:
    print("💡 Add GEMINI_API_KEY to .env for AI recommendations")
````
