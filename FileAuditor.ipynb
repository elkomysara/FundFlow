### **Cell 3: Directory Scanner Function**


## 🔍 Directory Scanning Function

This function recursively scans the project directory using modern `pathlib` for cross-platform compatibility. It returns separate lists for directories and files with full path information.


```python

def scan_project_directory(root_path: Path) -> Tuple[List[Path], List[Path]]:
    """
    Recursively scan directory and collect all folders and files.
    
    Args:
        root_path: Path object pointing to the root directory
        
    Returns:
        Tuple of (directories_list, files_list) with Path objects
    """
    directories = []
    files = []
    
    try:
        # Add the root directory itself
        directories.append(root_path)
        
        # Recursively scan all items
        for item in root_path.rglob("*"):
            if item.is_dir():
                directories.append(item)
            else:
                files.append(item)
                
    except PermissionError as e:
        print(f"⚠️ Permission denied accessing some directories: {e}")
    except Exception as e:
        print(f"❌ Error during scanning: {e}")
        
    return directories, files

# Perform the initial scan
print("🔍 Scanning project directory...")
all_directories, all_files = scan_project_directory(project_path)

print(f"📁 Found {len(all_directories)} directories")
print(f"📄 Found {len(all_files)} files")
print(f"📊 Total items: {len(all_directories) + len(all_files)}")
```

### **Cell 5: Exclusion Filter Function**
## 🔧 Exclusion Filter Implementation

This function applies the exclusion patterns using `fnmatch` for powerful wildcard matching. It checks both the item name and relative path against all patterns.


```python

def should_exclude_item(item_path: Path, exclusion_patterns: List[str], root_path: Path) -> bool:
    """
    Determine if an item should be excluded based on patterns.
    
    Args:
        item_path: Path to check for exclusion
        exclusion_patterns: List of exclusion patterns (supports wildcards)
        root_path: Root path for relative path calculation
        
    Returns:
        True if item should be excluded, False otherwise
    """
    try:
        # Get relative path from root
        relative_path = item_path.relative_to(root_path)
        
        # Check against each exclusion pattern
        for pattern in exclusion_patterns:
            # Check the item name
            if fnmatch.fnmatch(item_path.name, pattern):
                return True
            
            # Check the relative path
            if fnmatch.fnmatch(str(relative_path), pattern):
                return True
            
            # Check each part of the path
            for part in relative_path.parts:
                if fnmatch.fnmatch(part, pattern):
                    return True
                    
    except ValueError:
        # Item is not relative to root_path
        return False
        
    return False

def apply_exclusions(directories: List[Path], files: List[Path], 
                    dir_patterns: List[str], file_patterns: List[str], 
                    root_path: Path) -> Tuple[List[Path], List[Path]]:
    """
    Apply exclusion patterns to filter directories and files.
    
    Returns:
        Tuple of (included_directories, included_files)
    """
    included_dirs = []
    included_files = []
    
    # Filter directories
    for directory in directories:
        if not should_exclude_item(directory, dir_patterns, root_path):
            included_dirs.append(directory)
    
    # Filter files (check against both directory and file patterns)
    all_patterns = dir_patterns + file_patterns
    for file in files:
        if not should_exclude_item(file, all_patterns, root_path):
            # Also check if the file's parent directory is excluded
            parent_excluded = False
            for parent in file.parents:
                if should_exclude_item(parent, dir_patterns, root_path):
                    parent_excluded = True
                    break
            
            if not parent_excluded:
                included_files.append(file)
    
    return included_dirs, included_files

# Apply exclusions
print("🔧 Applying exclusion filters...")
included_directories, included_files = apply_exclusions(
    all_directories, all_files, 
    EXCLUDE_DIRECTORIES, EXCLUDE_FILES, 
    project_path
)

# Display results
total_excluded_dirs = len(all_directories) - len(included_directories)
total_excluded_files = len(all_files) - len(included_files)

print(f"✅ Filtering complete!")
print(f"📁 Directories: {len(all_directories)} → {len(included_directories)} (excluded: {total_excluded_dirs})")
print(f"📄 Files: {len(all_files)} → {len(included_files)} (excluded: {total_excluded_files})")
print(f"📊 Total included: {len(included_directories) + len(included_files)} items")
```


### **Cell 7: Tree Structure Visualization**
## 🌳 Project Structure Tree

Generate a beautiful ASCII tree visualization of your project structure. This creates a hierarchical view that's easy to understand and perfect for documentation.


```python

def build_tree_structure(directories: List[Path], files: List[Path], root_path: Path) -> dict:
    """
    Build a nested dictionary representing the directory tree structure.
    
    Returns:
        Nested dictionary where keys are names and values are either:
        - dict (for directories with contents)
        - None (for files)
        - {} (for empty directories)
    """
    tree = {}
    
    # Combine all items and sort by path
    all_items = [(d, True) for d in directories] + [(f, False) for f in files]
    all_items.sort(key=lambda x: str(x[0].relative_to(root_path)))
    
    for item_path, is_dir in all_items:
        try:
            relative_path = item_path.relative_to(root_path)
            parts = relative_path.parts
            
            if not parts:  # Root directory
                continue
                
            current_level = tree
            
            # Navigate through the path parts
            for i, part in enumerate(parts):
                is_last_part = (i == len(parts) - 1)
                
                if is_last_part:
                    # Last part - set as file (None) or directory ({})
                    if part not in current_level:
                        current_level[part] = {} if is_dir else None
                else:
                    # Intermediate directory
                    if part not in current_level:
                        current_level[part] = {}
                    current_level = current_level[part]
                    
        except ValueError:
            continue  # Skip items not under root_path
            
    return tree

def print_tree(tree_dict: dict, prefix: str = "", is_last: bool = True) -> List[str]:
    """
    Recursively generate tree lines with proper ASCII art.
    
    Returns:
        List of formatted tree lines
    """
    lines = []
    
    # Sort items: directories first, then files, both alphabetically
    items = sorted(tree_dict.items(), key=lambda x: (x[1] is None, x[0].lower()))
    
    for i, (name, subtree) in enumerate(items):
        is_last_item = (i == len(items) - 1)
        
        # Choose the appropriate connector
        if is_last_item:
            connector = "└── "
            extension = "    "
        else:
            connector = "├── "
            extension = "│   "
        
        # Add appropriate icon
        if subtree is None:  # File
            icon = "📄 "
        else:  # Directory
            icon = "📁 "
            
        lines.append(f"{prefix}{connector}{icon}{name}")
        
        # Recursively process subdirectories
        if isinstance(subtree, dict) and subtree:
            lines.extend(print_tree(subtree, prefix + extension, is_last_item))
    
    return lines
# Generate and display the tree
print("🌳 Generating project structure tree...")
tree_structure = build_tree_structure(included_directories, included_files, project_path)

print("\n" + "="*70)
print("🌳 PROJECT STRUCTURE")
print("="*70)
print(f"📁 {project_path.name}/")

tree_lines = print_tree(tree_structure)
for line in tree_lines:
    print(line)

print(f"\n📊 Structure summary: {len(included_directories)} directories, {len(included_files)} files")

```

## **📝 Enhanced Cell 10: File Content Display with Export Progress Tracking**

## 📖 Silent File Export with Path Tracking

This optimized version exports all file contents to a text file while displaying **only the essential information** in Jupyter - no content clutter, just clean progress tracking.

**Jupyter Output:**
- ✅ Export file location and name
- 📋 Numbered list of file paths being processed  
- 📊 Final summary with success/failure counts
- 🎯 **No file content displayed** - everything goes to the text file

**Text File Contains:**
- Complete file headers with metadata
- Full file contents (with optional line limits)
- Comprehensive export summary


```python

from datetime import datetime
from pathlib import Path
from typing import List, Optional, Tuple

def format_file_content(file_path: Path, root_path: Path, 
                       max_lines: Optional[int] = None,
                       show_line_numbers: bool = True) -> Tuple[bool, str]:
    """
    Format file content as a string for export (no console display).
    """
    output_lines = []
    
    try:
        relative_path = file_path.relative_to(root_path)
        file_stats = file_path.stat()
        
        # File header with comprehensive metadata
        output_lines.extend([
            "="*80,
            f"📄 File: {relative_path}",
            f"📍 Full Path: {file_path}",
            f"📊 Size: {file_stats.st_size:,} bytes",
            f"🕒 Modified: {datetime.fromtimestamp(file_stats.st_mtime).strftime('%Y-%m-%d %H:%M:%S')}",
            f"🏷️  Extension: {file_path.suffix or 'No extension'}",
            "="*80
        ])
        
        # Read file content with proper error handling
        try:
            content = file_path.read_text(encoding='utf-8')
        except UnicodeDecodeError:
            try:
                content = file_path.read_text(encoding='cp1252')
            except UnicodeDecodeError:
                content = file_path.read_text(encoding='utf-8', errors='ignore')
        
        lines = content.splitlines()
        
        output_lines.extend([
            f"📏 Total Lines: {len(lines)}",
            "-"*80
        ])
        
        # Process content lines
        lines_to_show = lines[:max_lines] if max_lines else lines
        
        for i, line in enumerate(lines_to_show, 1):
            if show_line_numbers:
                formatted_line = f"{i:4d} │ {line}"
            else:
                formatted_line = line
            output_lines.append(formatted_line)
        
        # Add truncation notice if applicable
        if max_lines and len(lines) > max_lines:
            remaining = len(lines) - max_lines
            output_lines.extend([
                "",
                f"... ({remaining} more lines truncated for export)",
                f"Set max_lines_per_file=None to include all {len(lines)} lines"
            ])
        
        output_lines.append("")  # Empty line for separation
        return True, "\n".join(output_lines)
        
    except Exception as e:
        # Handle binary files and other errors
        if file_path.suffix.lower() in ['.exe', '.dll', '.bin', '.jpg', '.png', '.gif', '.pdf', '.zip', '.ico']:
            error_content = [
                "="*80,
                f"📄 File: {file_path.relative_to(root_path)}",
                "⚠️ Binary file detected - cannot display as text",
                f"📊 Size: {file_path.stat().st_size:,} bytes",
                f"🏷️  Type: {file_path.suffix.upper()} file",
                "="*80,
                "This appears to be a binary file (executable, image, archive, etc.)",
                ""
            ]
        else:
            error_content = [
                "="*80,
                f"📄 File: {file_path.relative_to(root_path)}",
                f"❌ Error reading file: {str(e)}",
                "="*80,
                ""
            ]
        
        return False, "\n".join(error_content)

def display_multiple_files(file_list: List[Path], root_path: Path, 
                          max_lines_per_file: Optional[int] = None,
                          show_line_numbers: bool = True,
                          export_to_file: bool = True,  # Changed default to True
                          export_filename: Optional[str] = None,
                          export_path: Optional[Path] = None,
                          silent_mode: bool = True) -> Optional[Path]:  # New parameter
    """
    Export file contents to text file with minimal Jupyter output.
    
    Args:
        file_list: List of file paths to process
        root_path: Project root path
        max_lines_per_file: Maximum lines per file (None for all)
        show_line_numbers: Whether to show line numbers in export
        export_to_file: Whether to export content to a text file
        export_filename: Custom filename for export
        export_path: Custom directory for export file
        silent_mode: If True, only show paths and summary (no content in Jupyter)
        
    Returns:
        Path to exported file if export_to_file=True, None otherwise
    """
    
    if not file_list:
        print("⚠️ No files selected for export.")
        return None
    
    if not export_to_file:
        print("⚠️ Export is disabled. Set export_to_file=True to create a text file.")
        return None
    
    # Determine export location and filename
    if export_path is None:
        export_path = root_path
    
    if export_filename is None:
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        export_filename = f"file_contents_{root_path.name}_{timestamp}.txt"
    
    # Ensure .txt extension
    if not export_filename.endswith('.txt'):
        export_filename += '.txt'
    
    export_file_path = Path(export_path) / export_filename
    
    # Create export header
    export_header = [
        "="*80,
        "📦 PROJECT FILE CONTENT EXPORT",
        "="*80,
        f"🕒 Export Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}",
        f"📁 Project Name: {root_path.name}",
        f"📍 Project Path: {root_path}",
        f"📊 Total Files Selected: {len(file_list)}",
        f"⚙️  Max Lines Per File: {'All lines' if max_lines_per_file is None else max_lines_per_file}",
        f"🔢 Line Numbers: {'Enabled' if show_line_numbers else 'Disabled'}",
        "="*80,
        "",
        ""
    ]
    
    # Write header to file
    try:
        export_file_path.write_text("\n".join(export_header), encoding='utf-8')
    except Exception as e:
        print(f"❌ Error creating export file: {e}")
        return None
    
    # Display minimal startup info
    print(f"📤 Exporting {len(file_list)} file(s) to: {export_file_path.name}")
    print(f"📂 Location: {export_file_path.parent}")
    print("\n📋 Processing files:")
    
    # Track results
    exported_files: List[Path] = []
    failed_files: List[str] = []
    
    # Process each file
    for file_index, file_path in enumerate(file_list, 1):
        relative_path = file_path.relative_to(root_path)
        
        try:
            # Format the file content (for export only)
            success, formatted_content = format_file_content(
                file_path, root_path, max_lines_per_file, show_line_numbers
            )
            
            # Append to export file
            with open(export_file_path, 'a', encoding='utf-8') as f:
                f.write(formatted_content + "\n")
            
            # Show clean progress in Jupyter (just the path)
            if success:
                print(f"   {file_index:3d}. ✅ {relative_path}")
                exported_files.append(file_path)
            else:
                print(f"   {file_index:3d}. ⚠️ {relative_path} (Binary/Error - noted in export)")
                exported_files.append(file_path)  # Still count as processed
            
        except Exception as e:
            print(f"   {file_index:3d}. ❌ {relative_path} - Export failed")
            failed_files.append(f"{relative_path}: {str(e)}")
    
    # Display final summary
    print(f"\n📊 Export Summary:")
    print(f"   ✅ Successfully exported: {len(exported_files)}/{len(file_list)} files")
    
    if failed_files:
        print(f"   ❌ Failed exports: {len(failed_files)}")
    
    # Add export file statistics
    try:
        file_size = export_file_path.stat().st_size
        print(f"   📄 Export file size: {file_size:,} bytes ({file_size / 1024:.1f} KB)")
        if file_size > 1024 * 1024:
            print(f"   💾 File size: {file_size / (1024 * 1024):.1f} MB")
    except:
        pass
    
    print(f"\n💾 Export completed: {export_file_path.absolute()}")
    
    # Append final summary to export file
    try:
        with open(export_file_path, 'a', encoding='utf-8') as f:
            summary_content = [
                "\n" + "="*80,
                "📊 EXPORT SUMMARY",
                "="*80,
                f"✅ Successfully exported: {len(exported_files)}/{len(file_list)} files",
                f"🕒 Export completed: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}",
                "",
                "📋 EXPORTED FILE PATHS:",
                "="*40
            ]
            
            for i, file_path in enumerate(exported_files, 1):
                relative_path = file_path.relative_to(root_path)
                summary_content.append(f"{i:3d}. {relative_path}")
                summary_content.append(f"     Full Path: {file_path}")
                summary_content.append("")
            
            if failed_files:
                summary_content.extend([
                    "❌ FAILED EXPORTS:",
                    "="*20
                ])
                for failed in failed_files:
                    summary_content.append(f"   • {failed}")
                summary_content.append("")
            
            f.write("\n".join(summary_content))
    except Exception as e:
        print(f"⚠️ Could not append summary to export file: {e}")
    
    return export_file_path

# ==========================================
# 🎯 CONFIGURE EXPORT OPTIONS  
# ==========================================

# Export settings (optimized for silent mode)
MAX_LINES_PER_FILE = None  # None = all lines, or set a number like 50, 100, 200
SHOW_LINE_NUMBERS = True   # True = show line numbers in export file
EXPORT_TO_FILE = True      # Must be True for this mode
EXPORT_FILENAME = None     # None = auto-generate, or specify like "my_export.txt"
EXPORT_DIRECTORY = None    # None = save in project root, or specify Path object

# ==========================================
# 🚀 EXECUTE SILENT EXPORT
# ==========================================

exported_file = display_multiple_files(
    file_list=selected_files,
    root_path=project_path, 
    max_lines_per_file=MAX_LINES_PER_FILE,
    show_line_numbers=SHOW_LINE_NUMBERS,
    export_to_file=EXPORT_TO_FILE,
    export_filename=EXPORT_FILENAME,
    export_path=EXPORT_DIRECTORY,
    silent_mode=True  # This enables the clean output mode
)

if exported_file:
    print(f"🎉 Ready! Open your export file: {exported_file.name}")
```

### **Cell 11: Advanced Content Search**
## 🔍 Advanced Content Search

Search for specific keywords or patterns across selected files. This is useful for finding specific functions, variables, or text patterns in your codebase.


```python

def search_in_files(file_list: List[Path], root_path: Path, 
                   search_term: str, case_sensitive: bool = False,
                   context_lines: int = 2) -> None:
    """
    Search for a term across multiple files and display matches with context.
    
    Args:
        file_list: List of files to search
        root_path: Project root path
        search_term: Term to search for
        case_sensitive: Whether search should be case-sensitive
        context_lines: Number of lines to show before/after each match
    """
    if not search_term.strip():
        print("⚠️ Please provide a search term.")
        return
    
    if not file_list:
        print("⚠️ No files selected for search.")
        return
    
    total_matches = 0
    files_with_matches = 0
    
    print(f"🔍 Searching for '{search_term}' in {len(file_list)} file(s)...")
    print(f"🔧 Case sensitive: {case_sensitive}")
    print(f"📝 Context lines: {context_lines}")
    print("="*80)
    
    for file_path in file_list:
        try:
            content = file_path.read_text(encoding='utf-8')
            lines = content.splitlines()
            
            matches_in_file = []
            search_target = search_term if case_sensitive else search_term.lower()
            
            # Find all matching lines
            for line_num, line in enumerate(lines, 1):
                line_to_search = line if case_sensitive else line.lower()
                if search_target in line_to_search:
                    matches_in_file.append((line_num, line))
            
            if matches_in_file:
                files_with_matches += 1
                total_matches += len(matches_in_file)
                
                relative_path = file_path.relative_to(root_path)
                print(f"\n📄 {relative_path}")
                print(f"🎯 Found {len(matches_in_file)} match(es)")
                print("-" * 60)
                
                for line_num, matching_line in matches_in_file:
                    # Show context lines
                    start_line = max(1, line_num - context_lines)
                    end_line = min(len(lines), line_num + context_lines)
                    
                    for i in range(start_line, end_line + 1):
                        if i <= len(lines):
                            prefix = ">>> " if i == line_num else "    "
                            print(f"{prefix}{i:4d} │ {lines[i-1]}")
                    print()
                    
        except UnicodeDecodeError:
            print(f"⚠️ Skipping binary file: {file_path.relative_to(root_path)}")
        except Exception as e:
            print(f"❌ Error searching {file_path.relative_to(root_path)}: {e}")
    
    print("="*80)
    print(f"✅ Search complete!")
    print(f"📊 Found {total_matches} total matches in {files_with_matches} file(s)")
    print("="*80)

# ===== CONFIGURE YOUR SEARCH HERE =====
search_keyword = "def "  # Modify this to search for specific terms
case_sensitive_search = False
context_lines_to_show = 2

# Perform the search
if search_keyword.strip():
    search_in_files(selected_files, project_path, search_keyword, 
                   case_sensitive_search, context_lines_to_show)
else:
    print("ℹ️ Set 'search_keyword' variable above to perform a search")
```

```python

def generate_audit_summary(root_path: Path, all_dirs: List[Path], all_files: List[Path],
                          included_dirs: List[Path], included_files: List[Path],
                          selected_files: List[Path]) -> str:
    """Generate a comprehensive audit summary."""
    
    summary_lines = [
        "="*80,
        "📊 PROJECT AUDIT SUMMARY",
        "="*80,
        f"📁 Project: {root_path.name}",
        f"📍 Location: {root_path}",
        f"🕒 Audit Date: {__import__('datetime').datetime.now().strftime('%Y-%m-%d %H:%M:%S')}",
        "",
        "📈 SCAN RESULTS:",
        f"   📁 Total directories found: {len(all_dirs):,}",
        f"   📄 Total files found: {len(all_files):,}",
        f"   📊 Total items scanned: {len(all_dirs) + len(all_files):,}",
        "",
        "🔧 FILTERING RESULTS:",
        f"   📁 Directories included: {len(included_dirs):,} ({len(all_dirs) - len(included_dirs):,} excluded)",
        f"   📄 Files included: {len(included_files):,} ({len(all_files) - len(included_files):,} excluded)",
        f"   📊 Total items included: {len(included_dirs) + len(included_files):,}",
        "",
        "🎯 INSPECTION RESULTS:",
        f"   📖 Files selected for content inspection: {len(selected_files)}",
        "",
        "🏷️ FILE TYPE BREAKDOWN:",
    ]
    
    # Analyze file types
    file_types = {}
    total_size = 0
    
    for file_path in included_files:
        suffix = file_path.suffix.lower() or "no extension"
        file_size = file_path.stat().st_size
        
        if suffix not in file_types:
            file_types[suffix] = {"count": 0, "size": 0}
        
        file_types[suffix]["count"] += 1
        file_types[suffix]["size"] += file_size
        total_size += file_size
    
    # Sort by count
    sorted_types = sorted(file_types.items(), key=lambda x: x[1]["count"], reverse=True)
    
    for suffix, stats in sorted_types[:10]:  # Show top 10
        percentage = (stats["count"] / len(included_files)) * 100
        size_mb = stats["size"] / (1024 * 1024)
        summary_lines.append(f"   {suffix:15} : {stats['count']:4d} files ({percentage:5.1f}%) - {size_mb:6.1f} MB")
    
    if len(sorted_types) > 10:
        summary_lines.append(f"   ... and {len(sorted_types) - 10} more file types")
    
    summary_lines.extend([
        "",
        f"💾 TOTAL PROJECT SIZE: {total_size / (1024 * 1024):.1f} MB ({total_size:,} bytes)",
        "="*80
    ])
    
    return "\n".join(summary_lines)

def export_results(summary: str, tree_lines: List[str], export_path: Optional[str] = None):
    """Export audit results to a file."""
    
    if not export_path:
        export_path = f"project_audit_{project_path.name}_{__import__('datetime').datetime.now().strftime('%Y%m%d_%H%M%S')}.txt"
    
    export_content = [
        summary,
        "\n\n",
        "="*80,
        "🌳 PROJECT STRUCTURE",
        "="*80,
        f"📁 {project_path.name}/",
    ]
    
    export_content.extend(tree_lines)
    
    with open(export_path, 'w', encoding='utf-8') as f:
        f.write("\n".join(export_content))
    
    print(f"✅ Results exported to: {export_path}")
# Generate and display summary
audit_summary = generate_audit_summary(
    project_path, all_directories, all_files,
    included_directories, included_files, selected_files
)

print(audit_summary)

# Uncomment the line below to export results
# export_results(audit_summary, tree_lines)
print("\nℹ️ Uncomment the last line to export audit results to a file")
```

### **Cell 14: Comprehensive Data File Analysis**


## 📋 Part 4.1: Discover All CSV and Excel Files

This cell scans your included files for all data files and displays them in a numbered list with comprehensive metadata. Use these numbers in the next cell to select specific files for detailed analysis.

**Supported Formats:**
- 📊 `.csv` files (with encoding detection)
- 📗 `.xlsx` files (Excel 2007+)  
- 📘 `.xls` files (Excel 97-2003)


```python
def discover_and_display_data_files(file_list: List[Path], root_path: Path) -> List[Path]:
    """
    Discover all CSV and Excel files and display them with comprehensive metadata.
    
    Returns:
        Sorted list of discovered data file paths
    """
    
    # Define supported data file extensions
    data_extensions = {'.csv', '.xlsx', '.xls'}
    
    # Filter for data files
    data_files = [
        f for f in file_list 
        if f.suffix.lower() in data_extensions
    ]
    
    if not data_files:
        print("⚠️ No CSV or Excel files found in the included files.")
        print("💡 Check your exclusion filters - data files might be excluded.")
        print("📂 Common data file locations: data/, exports/, reports/, csv/")
        return []
    
    # Sort files for consistent numbering
    data_files.sort(key=lambda x: str(x.relative_to(root_path)))
    
    print("="*90)
    print(f"📊 DISCOVERED DATA FILES - Total: {len(data_files)}")
    print("="*90)
    print("💡 Use the index numbers below in the next cell for selective analysis\n")
    
    # Display each file with comprehensive metadata
    for index, file_path in enumerate(data_files, 1):
        relative_path = file_path.relative_to(root_path)
        file_extension = file_path.suffix.lower()
        
        try:
            file_stats = file_path.stat()
            file_size = file_stats.st_size
            modified_time = datetime.fromtimestamp(file_stats.st_mtime).strftime('%Y-%m-%d %H:%M')
            
            # Choose appropriate icon based on file type
            if file_extension == '.csv':
                icon = "📊"
                file_type = "CSV"
            elif file_extension == '.xlsx':
                icon = "📗"
                file_type = "Excel (XLSX)"
            elif file_extension == '.xls':
                icon = "📘"
                file_type = "Excel (XLS)"
            else:
                icon = "📄"
                file_type = "Data File"
            
            print(f"[{index:2d}] {icon} {relative_path}")
            print(f"     📍 Path: {file_path}")
            print(f"     📦 Type: {file_type}")
            print(f"     💾 Size: {file_size:,} bytes ({file_size / 1024:.1f} KB)")
            print(f"     🕒 Modified: {modified_time}")
            
            # Try to get quick preview info for Excel files
            if file_extension in ['.xlsx', '.xls']:
                try:
                    excel_file = pd.ExcelFile(file_path)
                    sheet_count = len(excel_file.sheet_names)
                    if sheet_count > 1:
                        print(f"     📚 Sheets: {sheet_count} ({', '.join(excel_file.sheet_names[:3])}{'...' if sheet_count > 3 else ''})")
                    else:
                        print(f"     📄 Sheets: 1 ('{excel_file.sheet_names[0]}')")
                except:
                    print(f"     📄 Sheets: Unable to read")
            
            print()
            
        except Exception as e:
            print(f"[{index:2d}] {icon} {relative_path}")
            print(f"     ⚠️ Warning: Could not read file metadata - {str(e)}")
            print()
    
    print("="*90)
    print(f"📋 Discovery Summary: {len(data_files)} data file(s) ready for analysis")
    print("📝 Proceed to the next cell to analyze these files")
    print("="*90)
    
    return data_files

# ==========================================
# 🚀 DISCOVER ALL DATA FILES
# ==========================================

print("🔍 Scanning for CSV and Excel files...\n")
discovered_data_files = discover_and_display_data_files(included_files, project_path)

if discovered_data_files:
    print(f"\n✅ Discovery complete! Found {len(discovered_data_files)} data file(s)")
    print("🎯 Ready for schema analysis in the next cell")
else:
    print("\n❌ No data files discovered")
    print("💡 Tips:")
    print("   - Check if data files are excluded by your filters")
    print("   - Ensure files have .csv, .xlsx, or .xls extensions")
    print("   - Verify files are in the included directories")
```

## 🔍 Part 4.2: Comprehensive Schema Analysis

```python
def analyze_data_file_schemas(available_files: List[Path], root_path: Path, 
                              selected_indices: Optional[List[int]] = None):
    """
    Comprehensive analysis of CSV and Excel file schemas.
    
    Args:
        available_files: List of discovered data files
        root_path: Project root path
        selected_indices: List of 1-based indices for specific files, or None for all
    """
    
    if not available_files:
        print("❌ No data files available for analysis.")
        print("💡 Run the previous cell first to discover data files.")
        return
    
    # Determine which files to analyze
    if selected_indices:
        files_to_analyze = []
        print(f"🎯 Selective Analysis Mode: {len(selected_indices)} file(s) selected")
        
        for idx in selected_indices:
            if 1 <= idx <= len(available_files):
                files_to_analyze.append(available_files[idx - 1])
                rel_path = available_files[idx - 1].relative_to(root_path)
                print(f"   ✅ [{idx}] {rel_path}")
            else:
                print(f"   ⚠️ Index {idx} is out of range (valid: 1-{len(available_files)})")
        
        if not files_to_analyze:
            print("❌ No valid files selected for analysis.")
            return
    else:
        files_to_analyze = available_files
        print(f"🎯 Complete Analysis Mode: All {len(available_files)} file(s)")
    
    print("\n" + "="*90)
    print("📊 DATA FILE SCHEMA ANALYSIS")
    print("="*90)
    
    analysis_stats = {
        'files_processed': 0,
        'sheets_analyzed': 0,
        'total_columns': 0,
        'successful_analyses': 0,
        'failed_analyses': 0
    }
    
    # Process each selected file
    for file_number, file_path in enumerate(files_to_analyze, 1):
        relative_path = file_path.relative_to(root_path)
        file_extension = file_path.suffix.lower()
        
        # Find original index in the full list
        original_index = available_files.index(file_path) + 1
        
        print(f"\n{'='*90}")
        print(f"📄 FILE [{file_number}/{len(files_to_analyze)}] - Original Index #{original_index}")
        print(f"📂 {relative_path}")
        print(f"{'='*90}")
        print(f"📍 Full Path: {file_path}")
        print(f"📦 Type: {file_extension.upper()}")
        print(f"💾 Size: {file_path.stat().st_size:,} bytes")
        
        analysis_stats['files_processed'] += 1
        
        try:
            # Analyze based on file type
            if file_extension == '.csv':
                schemas = analyze_csv_file(file_path)
            elif file_extension in ['.xlsx', '.xls']:
                schemas = analyze_excel_file(file_path)
            else:
                print(f"❌ Unsupported file type: {file_extension}")
                analysis_stats['failed_analyses'] += 1
                continue
            
            # Display results for each schema (CSV=1, Excel=multiple sheets)
            for schema in schemas:
                if 'error' in schema:
                    print(f"❌ Error: {schema['error']}")
                    analysis_stats['failed_analyses'] += 1
                else:
                    display_detailed_schema(schema)
                    analysis_stats['sheets_analyzed'] += 1
                    analysis_stats['total_columns'] += len(schema['columns'])
                    analysis_stats['successful_analyses'] += 1
                    
        except Exception as e:
            print(f"\n❌ CRITICAL ERROR analyzing file:")
            print(f"   {str(e)}")
            print(f"   💡 File may be corrupted, password-protected, or in an unsupported format")
            analysis_stats['failed_analyses'] += 1
    
    # Display comprehensive summary
    display_analysis_summary(analysis_stats, len(files_to_analyze), selected_indices is not None)

def analyze_csv_file(file_path: Path) -> List[Dict]:
    """Analyze CSV file with robust encoding detection."""
    
    # Windows-compatible encoding detection sequence
    encodings = ['utf-8', 'cp1252', 'latin1', 'iso-8859-1', 'utf-16']
    
    for encoding in encodings:
        try:
            # Read sample for type inference
            df_sample = pd.read_csv(file_path, encoding=encoding, nrows=10)
            
            # Estimate total rows efficiently
            try:
                with open(file_path, 'r', encoding=encoding) as f:
                    total_rows = sum(1 for _ in f) - 1  # Subtract header
            except:
                total_rows = "Unknown"
            
            return [{
                'sheet_name': None,
                'encoding': encoding,
                'total_rows': total_rows,
                'total_columns': len(df_sample.columns),
                'columns': [(i+1, col_name, str(dtype)) 
                           for i, (col_name, dtype) in enumerate(df_sample.dtypes.items())]
            }]
            
        except (UnicodeDecodeError, pd.errors.ParserError):
            continue
        except Exception as e:
            if encoding == encodings[-1]:  # Last attempt
                return [{'error': f"CSV parsing failed: {str(e)}"}]
            continue
    
    return [{'error': "Could not decode CSV with any supported encoding"}]

def analyze_excel_file(file_path: Path) -> List[Dict]:
    """Analyze Excel file with multi-sheet support."""
    
    schemas = []
    
    try:
        excel_file = pd.ExcelFile(file_path)
        sheet_names = excel_file.sheet_names
        
        print(f"📚 Workbook contains {len(sheet_names)} sheet(s): {', '.join(sheet_names)}")
        
        for sheet_name in sheet_names:
            try:
                # Read sample for type inference
                df_sample = pd.read_excel(file_path, sheet_name=sheet_name, nrows=10)
                
                # Estimate total rows
                try:
                    df_full = pd.read_excel(file_path, sheet_name=sheet_name, usecols=[0])
                    total_rows = len(df_full)
                except:
                    total_rows = "Unknown"
                
                schemas.append({
                    'sheet_name': sheet_name,
                    'encoding': None,
                    'total_rows': total_rows,
                    'total_columns': len(df_sample.columns),
                    'columns': [(i+1, col_name, str(dtype)) 
                               for i, (col_name, dtype) in enumerate(df_sample.dtypes.items())]
                })
                
            except Exception as e:
                schemas.append({
                    'sheet_name': sheet_name,
                    'error': f"Sheet analysis failed: {str(e)}"
                })
        
    except Exception as e:
        return [{'error': f"Excel file access failed: {str(e)}"}]
    
    return schemas

def display_detailed_schema(schema: Dict):
    """Display comprehensive schema information."""
    
    print(f"\n📋 SCHEMA DETAILS:")
    
    if schema.get('sheet_name'):
        print(f"📄 Sheet Name: '{schema['sheet_name']}'")
    
    if schema.get('encoding'):
        print(f"🔤 File Encoding: {schema['encoding']}")
    
    print(f"📏 Estimated Rows: {schema['total_rows']:,}" if isinstance(schema['total_rows'], int) else f"📏 Estimated Rows: {schema['total_rows']}")
    print(f"📊 Total Columns: {schema['total_columns']}")
    
    print(f"\n📋 COLUMN SCHEMA ({schema['total_columns']} columns):")
    print("-" * 80)
    
    if not schema['columns']:
        print("   (No columns detected - file may be empty)")
        return
    
    # Display formatted column list
    for col_number, col_name, col_type in schema['columns']:
        # Clean up pandas data types for readability
        clean_type = (col_type
                     .replace('int64', 'Integer')
                     .replace('int32', 'Integer') 
                     .replace('float64', 'Float')
                     .replace('float32', 'Float')
                     .replace('object', 'Text/String')
                     .replace('bool', 'Boolean')
                     .replace('datetime64[ns]', 'DateTime')
                     .replace('category', 'Category'))
        
        print(f"   {col_number:3d}. {col_name:<45} → {clean_type}")

def display_analysis_summary(stats: Dict, files_count: int, is_selective: bool):
    """Display comprehensive analysis summary."""
    
    print(f"\n{'='*90}")
    print("📈 ANALYSIS SUMMARY")
    print("="*90)
    print(f"🎯 Analysis Mode: {'Selective' if is_selective else 'Complete'}")
    print(f"📁 Files Processed: {stats['files_processed']}/{files_count}")
    print(f"📄 Sheets/Files Analyzed: {stats['sheets_analyzed']}")
    print(f"✅ Successful Analyses: {stats['successful_analyses']}")
    print(f"❌ Failed Analyses: {stats['failed_analyses']}")
    print(f"📋 Total Columns Discovered: {stats['total_columns']:,}")
    
    if stats['successful_analyses'] > 0:
        avg_columns = stats['total_columns'] / stats['successful_analyses']
        print(f"📊 Average Columns per Sheet: {avg_columns:.1f}")
    
    print("="*90)

# ==========================================
# 🎯 CONFIGURATION: SELECT YOUR FILES
# ==========================================

# OPTION 1: Analyze specific files by their index numbers
SELECTED_FILE_INDICES = [2, 3]  # ← Edit these numbers based on Cell 14 output

# OPTION 2: Analyze all files (comment out the line above and uncomment below)
# SELECTED_FILE_INDICES = None

# ==========================================
# 🚀 EXECUTE SCHEMA ANALYSIS  
# ==========================================

if not discovered_data_files:
    print("❌ No data files available for analysis.")
    print("💡 Please run Cell 14 first to discover data files.")
else:
    print("🚀 Starting comprehensive schema analysis...\n")
    
    analyze_data_file_schemas(
        available_files=discovered_data_files,
        root_path=project_path,
        selected_indices=SELECTED_FILE_INDICES
    )
```

This cell creates a **professional narrative report** that systematically describes your entire project in well-structured paragraphs. Perfect for documentation, project reviews, or stakeholder presentations.

```python
# =========================
# 📄 Cell 16: Project Report
# =========================

from datetime import datetime
from pathlib import Path
from typing import List, Optional, Dict
from collections import Counter
import pandas as pd

def analyze_project_structure(root_path: Path, directories: List[Path], files: List[Path]) -> Dict:
    """Analyze project structure and calculate key metrics."""
    
    # Calculate total size
    total_size = 0
    for file_path in files:
        try:
            total_size += file_path.stat().st_size
        except Exception:
            continue
    
    # Analyze file types
    extensions = Counter()
    for file_path in files:
        ext = file_path.suffix.lower() or 'no_extension'
        extensions[ext] += 1
    
    # Analyze directory structure
    max_depth = 0
    top_level_dirs = []
    
    for directory in directories:
        try:
            relative_parts = directory.relative_to(root_path).parts
            if len(relative_parts) == 1:  # Top-level directory
                top_level_dirs.append(directory.name)
            max_depth = max(max_depth, len(relative_parts))
        except ValueError:
            continue
    
    # Files per directory analysis
    dir_file_counts = Counter()
    for file_path in files:
        dir_file_counts[file_path.parent] += 1
    
    return {
        'total_size_bytes': total_size,
        'total_size_mb': total_size / (1024 * 1024),
        'file_extensions': extensions,
        'max_depth': max_depth,
        'top_level_dirs': sorted(top_level_dirs),
        'dir_file_counts': dir_file_counts,
        'most_populated_dir': max(dir_file_counts.items(), key=lambda x: x[1]) if dir_file_counts else None
    }

def analyze_data_files_comprehensive(data_files: List[Path], root_path: Path) -> List[Dict]:
    """Perform comprehensive analysis of CSV and Excel files."""
    
    data_file_analyses = []
    
    for file_path in data_files:
        relative_path = file_path.relative_to(root_path)
        analysis = {
            'path': file_path,
            'relative_path': relative_path,
            'name': file_path.name,
            'extension': file_path.suffix.lower(),
            'size_bytes': 0,
            'location_description': '',
            'sheets': [],
            'error': None
        }
        
        # Get file stats
        try:
            stats = file_path.stat()
            analysis['size_bytes'] = stats.st_size
            analysis['modified_date'] = datetime.fromtimestamp(stats.st_mtime).strftime('%B %d, %Y')
        except Exception:
            analysis['modified_date'] = 'Unknown'
        
        # Create location description
        parent_path = relative_path.parent
        if str(parent_path) == '.':
            analysis['location_description'] = "in the project root directory"
        else:
            path_parts = parent_path.parts
            if len(path_parts) == 1:
                analysis['location_description'] = f"within the '{path_parts[0]}' directory"
            else:
                analysis['location_description'] = (
                    f"within the '{path_parts[0]}' directory structure, "
                    f"specifically in the '{parent_path}' subdirectory"
                )
        
        # Analyze file schema
        try:
            if analysis['extension'] == '.csv':
                # Try multiple encodings for CSV
                encodings = ['utf-8', 'cp1252', 'latin1', 'iso-8859-1']
                df = None
                used_encoding = None
                
                for encoding in encodings:
                    try:
                        df = pd.read_csv(file_path, encoding=encoding, nrows=10)
                        used_encoding = encoding
                        break
                    except Exception:
                        continue
                
                if df is not None:
                    # Estimate row count
                    try:
                        with open(file_path, 'r', encoding=used_encoding) as f:
                            estimated_rows = sum(1 for _ in f) - 1  # Subtract header
                    except Exception:
                        estimated_rows = 'Unknown'
                    
                    columns = [
                        (
                            col_name,
                            str(dtype)
                            .replace('int64', 'Integer')
                            .replace('int32', 'Integer')
                            .replace('float64', 'Float')
                            .replace('float32', 'Float')
                            .replace('object', 'Text/String')
                            .replace('bool', 'Boolean')
                            .replace('datetime64[ns]', 'DateTime')
                        )
                        for col_name, dtype in df.dtypes.items()
                    ]
                    
                    analysis['sheets'].append({
                        'name': None,
                        'encoding': used_encoding,
                        'columns': columns,
                        'estimated_rows': estimated_rows
                    })
                else:
                    analysis['error'] = "Could not decode CSV file with supported encodings"
            
            elif analysis['extension'] in ['.xlsx', '.xls']:
                excel_file = pd.ExcelFile(file_path)
                
                for sheet_name in excel_file.sheet_names:
                    try:
                        df = pd.read_excel(file_path, sheet_name=sheet_name, nrows=10)
                        
                        # Try to get row count for first sheet
                        if sheet_name == excel_file.sheet_names[0]:
                            try:
                                full_df = pd.read_excel(file_path, sheet_name=sheet_name, usecols=[0])
                                estimated_rows = len(full_df)
                            except Exception:
                                estimated_rows = 'Unknown'
                        else:
                            estimated_rows = 'Unknown'
                        
                        columns = [
                            (
                                col_name,
                                str(dtype)
                                .replace('int64', 'Integer')
                                .replace('int32', 'Integer')
                                .replace('float64', 'Float')
                                .replace('float32', 'Float')
                                .replace('object', 'Text/String')
                                .replace('bool', 'Boolean')
                                .replace('datetime64[ns]', 'DateTime')
                            )
                            for col_name, dtype in df.dtypes.items()
                        ]
                        
                        analysis['sheets'].append({
                            'name': sheet_name,
                            'encoding': None,
                            'columns': columns,
                            'estimated_rows': estimated_rows
                        })
                    except Exception as e:
                        analysis['sheets'].append({
                            'name': sheet_name,
                            'error': f"Could not analyze sheet: {str(e)}",
                            'columns': []
                        })
        
        except Exception as e:
            analysis['error'] = f"File analysis failed: {str(e)}"
        
        data_file_analyses.append(analysis)
    
    return data_file_analyses

def get_git_info(root_path: Path) -> Dict:
    """Extract Git repository information if available."""
    git_dir = root_path / '.git'
    
    if not git_dir.exists():
        return {'has_git': False}
    
    try:
        import subprocess
        
        # Get current branch
        branch = subprocess.check_output(
            ['git', 'rev-parse', '--abbrev-ref', 'HEAD'],
            cwd=root_path,
            stderr=subprocess.DEVNULL
        ).decode().strip()
        
        # Get last commit
        last_commit = subprocess.check_output(
            ['git', 'log', '-1', '--oneline'],
            cwd=root_path,
            stderr=subprocess.DEVNULL
        ).decode().strip()
        
        return {
            'has_git': True,
            'branch': branch,
            'last_commit': last_commit
        }
    except Exception:
        return {'has_git': True, 'error': 'Could not read Git info'}

def generate_narrative_report(
    root_path: Path, 
    structure_analysis: Dict,
    directories: List[Path], 
    files: List[Path],
    content_export_file: Optional[Path],
    data_file_analyses: List[Dict],
    git_info: Optional[Dict] = None
) -> str:
    """Generate the comprehensive narrative report."""
    
    report_sections: List[str] = []
    
    # === HEADER ===
    report_sections.extend([
        "=" * 80,
        "COMPREHENSIVE PROJECT ANALYSIS REPORT",
        f"Project: {root_path.name}",
        "=" * 80,
        f"Generated: {datetime.now().strftime('%A, %B %d, %Y at %I:%M %p')}",
        f"Project Location: {root_path}",
        "",
        ""
    ])
    
    # === EXECUTIVE SUMMARY ===
    report_sections.extend([
        "EXECUTIVE SUMMARY",
        "=" * 50,
        ""
    ])
    
    top_extensions = structure_analysis['file_extensions'].most_common(3)
    ext_summary = ", ".join([f"{ext} ({count} files)" for ext, count in top_extensions])
    
    summary_text = (
        f"This comprehensive analysis examines the '{root_path.name}' project, a "
        f"{'sophisticated multi-layered' if structure_analysis['max_depth'] > 3 else 'well-organized'} "
        f"software project located at {root_path}. The project contains {len(files):,} files "
        f"distributed across {len(directories):,} directories, with a total storage footprint of "
        f"{structure_analysis['total_size_mb']:.2f} MB ({structure_analysis['total_size_bytes']:,} bytes). "
        f"The codebase demonstrates technological diversity with {len(structure_analysis['file_extensions'])} "
        f"distinct file types, primarily consisting of {ext_summary}."
    )
    
    if data_file_analyses:
        summary_text += (
            f" The project incorporates {len(data_file_analyses)} data files in CSV and Excel formats, "
            f"indicating data-driven functionality and analytical capabilities."
        )
    
    if content_export_file:
        summary_text += (
            " A comprehensive content export has been generated for detailed code review "
            "and documentation purposes."
        )
    
    report_sections.extend([summary_text, "", ""])
    
    # === PROJECT STRUCTURE ANALYSIS ===
    report_sections.extend([
        "PROJECT STRUCTURE AND ORGANIZATION",
        "=" * 50,
        ""
    ])
    
    top_dirs_display = ', '.join([f"'{d}'" for d in structure_analysis['top_level_dirs'][:8]])
    if len(structure_analysis['top_level_dirs']) > 8:
        top_dirs_display += '...'
    
    structure_text = (
        f"The project exhibits a hierarchical structure with a maximum directory depth of "
        f"{structure_analysis['max_depth']} levels, demonstrating "
        f"{'sophisticated architectural organization' if structure_analysis['max_depth'] > 4 else 'clear structural organization'}. "
        f"The root directory '{root_path.name}' contains {len(structure_analysis['top_level_dirs'])} "
        f"primary organizational divisions: {top_dirs_display}. "
    )
    
    if structure_analysis['most_populated_dir']:
        most_pop_dir, file_count = structure_analysis['most_populated_dir']
        relative_most_pop = most_pop_dir.relative_to(root_path) if most_pop_dir != root_path else Path('.')
        structure_text += (
            f"The most populated directory is '{relative_most_pop}' with {file_count} files, "
            "serving as a central hub for project content."
        )
    
    report_sections.extend([structure_text, "", ""])
    
    # === VERSION CONTROL STATUS (Git) ===
    if git_info and git_info.get('has_git'):
        report_sections.extend([
            "VERSION CONTROL STATUS",
            "=" * 50,
            ""
        ])
        
        if 'error' in git_info:
            git_text = (
                "This project appears to use Git version control, but repository details "
                "could not be fully determined."
            )
        else:
            git_text = (
                f"This project uses Git version control. Current branch: '{git_info.get('branch', 'unknown')}'. "
                f"Last commit: {git_info.get('last_commit', 'unknown')}. This enables collaborative development, "
                "clear change history, and structured deployment workflows."
            )
        
        report_sections.extend([git_text, "", ""])
    
    # === CONTENT EXPORT DOCUMENTATION ===
    if content_export_file:
        report_sections.extend([
            "CONTENT EXPORT DOCUMENTATION",
            "=" * 50,
            ""
        ])
        
        try:
            export_size = content_export_file.stat().st_size
            export_size_kb = export_size / 1024
        except Exception:
            export_size = 0
            export_size_kb = 0
        
        content_text = (
            f"A consolidated content export has been generated and saved as '{content_export_file.name}', "
            f"located at {content_export_file.parent}. This comprehensive document captures the textual "
            "content of selected project files, totaling "
            f"{export_size_kb:.1f} KB ({export_size:,} bytes) of extracted source code and documentation. "
            "The export file serves as a complete snapshot of the project's human-readable assets, suitable "
            "for code review, offline analysis, or archival purposes. The full path to this export is: "
            f"{content_export_file}."
        )
        
        report_sections.extend([content_text, "", ""])
    
    # === DATA FILES ANALYSIS ===
    if data_file_analyses:
        report_sections.extend([
            "DATA ARCHITECTURE AND SCHEMA ANALYSIS",
            "=" * 50,
            ""
        ])
        
        csv_count = sum(1 for d in data_file_analyses if d['extension'] == '.csv')
        excel_count = sum(1 for d in data_file_analyses if d['extension'] in ['.xlsx', '.xls'])
        total_columns = sum(
            len(sheet.get('columns', []))
            for d in data_file_analyses
            for sheet in d['sheets']
        )
        
        data_intro = (
            f"The project incorporates a robust data infrastructure consisting of {len(data_file_analyses)} "
            f"structured data files ({csv_count} CSV files and {excel_count} Excel workbooks), "
            f"collectively containing approximately {total_columns} data columns. These files provide "
            "the analytical foundation and data storage capabilities for the project's operations."
        )
        
        report_sections.extend([data_intro, "", "DETAILED DATA FILE ANALYSIS:", ""])
        
        for i, analysis in enumerate(data_file_analyses, 1):
            file_section: List[str] = [
                f"Data File {i}: {analysis['name']}",
                "-" * 40
            ]
            
            file_description = (
                f"This {analysis['extension'].upper()} file is strategically positioned {analysis['location_description']}, "
                f"occupying {analysis['size_bytes']:,} bytes ({analysis['size_bytes']/1024:.1f} KB) of storage space. "
                f"Last modified on {analysis['modified_date']}, this file represents a key data asset within the project ecosystem."
            )
            
            file_section.extend([
                file_description,
                "",
                f"Complete file path: {analysis['path']}",
                f"Relative path: {analysis['relative_path']}",
                ""
            ])
            
            if analysis['error']:
                file_section.extend([
                    f"Analysis Note: {analysis['error']}",
                    ""
                ])
            elif analysis['sheets']:
                if analysis['extension'] == '.csv':
                    sheet = analysis['sheets'][0]
                    file_section.extend([
                        f"Schema Analysis (Encoding: {sheet.get('encoding', 'UTF-8')}):",
                        f"• Estimated rows: {sheet.get('estimated_rows', 'Unknown')}",
                        f"• Column count: {len(sheet.get('columns', []))}",
                        ""
                    ])
                    
                    if sheet.get('columns'):
                        file_section.append("Column Schema:")
                        for j, (col_name, col_type) in enumerate(sheet['columns'], 1):
                            file_section.append(f"   {j:2d}. {col_name:<30} → {col_type}")
                        file_section.append("")
                
                else:  # Excel file
                    file_section.extend([
                        f"Workbook Structure: {len(analysis['sheets'])} sheet(s)",
                        ""
                    ])
                    
                    for sheet in analysis['sheets']:
                        if 'error' in sheet:
                            file_section.append(f"• Sheet '{sheet['name']}': {sheet['error']}")
                        else:
                            file_section.append(
                                f"• Sheet '{sheet['name']}': {len(sheet.get('columns', []))} columns"
                            )
                            
                            if sheet.get('columns'):
                                for j, (col_name, col_type) in enumerate(sheet['columns'], 1):
                                    file_section.append(
                                        f"     {j:2d}. {col_name:<25} → {col_type}"
                                    )
                            
                            file_section.append("")
            
            report_sections.extend(file_section)
            report_sections.append("")
    
    # === CONCLUSION ===
    report_sections.extend([
        "ANALYSIS CONCLUSION",
        "=" * 50,
        ""
    ])
    
    conclusion = (
        f"This comprehensive analysis of the '{root_path.name}' project reveals a "
        f"{'sophisticated and well-architected' if structure_analysis['max_depth'] > 3 else 'well-organized and structured'} "
        f"codebase with {len(files):,} files spanning {len(directories):,} directories. "
        f"The project demonstrates {len(structure_analysis['file_extensions'])} distinct file types with a total size of "
        f"{structure_analysis['total_size_mb']:.2f} MB, indicating a substantial and diverse software project."
    )
    
    if data_file_analyses:
        conclusion += (
            f" The incorporation of {len(data_file_analyses)} structured data files with {total_columns} total columns "
            "demonstrates strong data-driven capabilities and analytical functionality."
        )
    
    if content_export_file:
        conclusion += (
            " The generated content export file provides comprehensive documentation for "
            "code review and project analysis."
        )
    
    conclusion += (
        f"\n\nThis report was generated on {datetime.now().strftime('%A, %B %d, %Y at %I:%M %p')} "
        "and represents a complete snapshot of the project structure and data architecture at this point in time."
    )
    
    report_sections.extend([
        conclusion,
        "",
        "=" * 80,
        "END OF REPORT",
        "=" * 80
    ])
    
    return "\n".join(report_sections)

# ==========================================
# 🎯 CONFIGURATION
# ==========================================

REPORT_FILENAME = None  # None = auto-generate, or specify custom name

# ==========================================
# 🚀 GENERATE COMPREHENSIVE REPORT
# ==========================================

print("📝 Generating comprehensive project narrative report...")

if 'project_path' not in globals():
    print("❌ Error: project_path not found. Please run the project configuration cell first.")
else:
    # Get variables with safe fallbacks
    dirs_to_analyze = globals().get('included_directories', [])
    files_to_analyze = globals().get('included_files', [])
    export_file = globals().get('exported_file', None)
    
    # Data files
    if 'discovered_data_files' in globals():
        data_files_list = discovered_data_files
    else:
        data_extensions = {'.csv', '.xlsx', '.xls'}
        data_files_list = [f for f in files_to_analyze if f.suffix.lower() in data_extensions]
    
    if not dirs_to_analyze or not files_to_analyze:
        print("⚠️ Warning: No directories or files found. Please run the project scanning cells first.")
    
    # Analyze project structure
    print("🔍 Analyzing project structure...")
    structure_stats = analyze_project_structure(project_path, dirs_to_analyze, files_to_analyze)
    
    # Analyze data files
    print("📊 Analyzing data files...")
    data_analyses = analyze_data_files_comprehensive(data_files_list, project_path)
    
    # Git info
    print("🔧 Checking Git repository status...")
    git_info = get_git_info(project_path)
    
    # Generate narrative report
    print("✍️ Generating narrative report...")
    report_content = generate_narrative_report(
        project_path,
        structure_stats,
        dirs_to_analyze,
        files_to_analyze,
        export_file,
        data_analyses,
        git_info=git_info
    )
    
    # Determine filename
    if REPORT_FILENAME is None:
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        report_filename = f"PROJECT_REPORT_{project_path.name}_{timestamp}.txt"
    else:
        report_filename = REPORT_FILENAME
        if not report_filename.endswith('.txt'):
            report_filename += '.txt'
    
    report_path = project_path / report_filename
    
    # Save report
    try:
        report_path.write_text(report_content, encoding='utf-8')
        
        print("\n" + "=" * 80)
        print("✅ PROJECT NARRATIVE REPORT GENERATED SUCCESSFULLY!")
        print("=" * 80)
        print(f"📄 Report Name: {report_filename}")
        print(f"📂 Location: {report_path.parent}")
        print(f"📍 Full Path: {report_path.absolute()}")
        
        report_size = report_path.stat().st_size
        print(f"📊 Report Size: {report_size:,} bytes ({report_size / 1024:.1f} KB)")
        
        report_lines = report_content.split('\n')
        print(f"📏 Total Lines: {len(report_lines):,}")
        
        print("\n📖 Report Preview (first 15 lines):")
        print("-" * 60)
        for line in report_lines[:15]:
            print(line)
        print("...")
        print("-" * 60)
        
        print(f"\n💡 Open '{report_filename}' to read the complete analysis.")
        print("🎯 The report includes structure, statistics, Git status, content export, and data schemas.")
        
    except Exception as e:
        print(f"❌ Error saving report: {e}")
```
